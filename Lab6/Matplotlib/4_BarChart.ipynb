```python
import matplotlib.pyplot as plt
```

## Sample Data

```python
# Years from 2000 to 2020 where both actors had direct releases (16 elements)
years = [2000, 2001, 2002, 2003, 2004, 2006, 2007, 2008, 2010, 2011, 2012, 2014, 2015, 2016, 2017, 2018]
```

```python
# Highest-grossing movie names per year for Shah Rukh Khan (16 elements)
srk_movie = [
    "Mohabbatein", "Kabhi Khushi Kabhie Gham...", "Devdas", "Kal Ho Naa Ho", 
    "Veer-Zaara", "Kabhi Alvida Naa Kehna", "Om Shanti Om", "Rab Ne Bana Di Jodi", 
    "My Name Is Khan", "Don 2", "Jab Tak Hai Jaan", "Happy New Year", 
    "Dilwale", "Fan", "Raees", "Zero"
]

# Worldwide gross revenue (in INR Crores) for Shah Rukh Khan's movies (16 elements)
srk_revenue = [90, 136, 100, 86, 94, 113, 150, 157, 223, 203, 211, 408, 372, 188, 272, 186]
```

```python
# Highest-grossing movie names per year for Salman Khan (16 elements)
salman_movie = [
    "Har Dil Jo Pyar Karega", "Chori Chori Chupke Chupke", "Hum Tumhare Hain Sanam", "Baghban", 
    "Mujhse Shaadi Karogi", "Jaan-E-Mann", "Partner", "God Tussi Great Ho", 
    "Dabangg", "Bodyguard", "Ek Tha Tiger", "Kick", 
    "Bajrangi Bhaijaan", "Sultan", "Tiger Zinda Hai", "Race 3"
]

# Worldwide gross revenue (in INR Crores) for Salman Khan's movies (16 elements)
salman_revenue = [96, 43, 35, 43, 56, 44, 103, 20, 221, 234, 335, 402, 969, 589, 565, 303]
```

### Making Data More Detailed for Plotting

```python
salman_data = list()

for name , year in zip(salman_movie , years):
    salman_data.append(f"{name} ({year})")
```

```python
srk_data = list()

for name,year in zip(srk_movie , years):
    srk_data.append(f"{name} ({year})")
```

## Vertical Bar Chart
- ### X-Axis -> Category
- ### Y-Axis -> Data

```python
plt.bar(salman_data , salman_revenue , color = "purple")
plt.xticks(ticks = salman_data , rotation = 90)
plt.title("Salman's Revenue")
plt.ylabel("Revenue (INR Cr.)")
plt.xlabel("Movies")

# To Get Exact Revenue on the top of bars
# We'll use plt.text(x-pos , y-pos , text , alignment )

# plt.text(salman_data[0] , salman_revenue[0] + 10 , str(salman_revenue[0]) , ha = "center")
# Using Above we can print values for all

for i in range(len(salman_data)):
    plt.text(i , salman_revenue[i] + 10 , str(salman_revenue[i]) , ha = "center")  
  
# Set Limits to X and Y-Axis
# plt.xlim(lower , upper )
# plt.ylim(lower , upper )

# max. salman_revenue = 969 , yLowerLimit = 0 , yUpperLimit = 969+100 = 1069
plt.ylim(0 , max(salman_revenue) + 100 )  

plt.show()
```

## Horizontal Bar Chart
- ### X-Axis -> Data
- ### Y-Axis -> Category

```python
plt.barh(srk_data , srk_revenue)
plt.title("SRK's Revenue")
plt.xlabel("Revenue (INR Cr.)")
plt.ylabel("Movies")

for i in range(len(srk_revenue)):
    # pass
    plt.text(srk_revenue[i] + 5, i , str(srk_revenue[i]) , ha = "left")


# maximum of srk_revenue = 408 , xLowerLimit = 0 , xUpperLimit = 408 + 100 = 508 
plt.xlim(0 , max(srk_revenue) + 100)


plt.show()
```
