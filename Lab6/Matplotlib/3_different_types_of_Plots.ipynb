```python
import matplotlib.pyplot as plt
```

### Sample Data

```python
# Years from 2000 to 2020 where both actors had direct releases (16 elements)
years = [2000, 2001, 2002, 2003, 2004, 2006, 2007, 2008, 2010, 2011, 2012, 2014, 2015, 2016, 2017, 2018]
```

```python
# Highest-grossing movie names per year for Shah Rukh Khan (16 elements)
srk_movie = [
    "Mohabbatein", "Kabhi Khushi Kabhie Gham...", "Devdas", "Kal Ho Naa Ho", 
    "Veer-Zaara", "Kabhi Alvida Naa Kehna", "Om Shanti Om", "Rab Ne Bana Di Jodi", 
    "My Name Is Khan", "Don 2", "Jab Tak Hai Jaan", "Happy New Year", 
    "Dilwale", "Fan", "Raees", "Zero"
]

# Worldwide gross revenue (in INR Crores) for Shah Rukh Khan's movies (16 elements)
srk_revenue = [90, 136, 100, 86, 94, 113, 150, 157, 223, 203, 211, 408, 372, 188, 272, 186]
```

```python
# Highest-grossing movie names per year for Salman Khan (16 elements)
salman_movie = [
    "Har Dil Jo Pyar Karega", "Chori Chori Chupke Chupke", "Hum Tumhare Hain Sanam", "Baghban", 
    "Mujhse Shaadi Karogi", "Jaan-E-Mann", "Partner", "God Tussi Great Ho", 
    "Dabangg", "Bodyguard", "Ek Tha Tiger", "Kick", 
    "Bajrangi Bhaijaan", "Sultan", "Tiger Zinda Hai", "Race 3"
]

# Worldwide gross revenue (in INR Crores) for Salman Khan's movies (16 elements)
salman_revenue = [96, 43, 35, 43, 56, 44, 103, 20, 221, 234, 335, 402, 969, 589, 565, 303]
```

### Different Types of Plot Methods

1. #### plt.plot(x, y) - Line plot
2. #### plt.bar(x, height) - Vertical Bar chart
3. #### plt.barh(y, width) - Horizontal Bar chart
4. #### plt.scatter(x, y) - Scatter plot
5. #### plt.pie(data) - Pie chart
6. #### plt.hist(data, bins) - Histogram
7. #### plt.boxplot(data) - Box plot

### plt.plot(x, y) - Line plot

```python
plt.plot(years , srk_revenue)
plt.xticks(ticks = years , rotation = 75)
plt.show()
```

### plt.bar(x, height) - Vertical Bar chart

```python
plt.bar(salman_movie , salman_revenue)
plt.xticks(rotation = 90)
plt.grid(axis = "y")
plt.show()
```

### plt.barh(y, width) - Horizontal Bar chart

```python
plt.barh( years , salman_revenue)
plt.xticks(rotation = 75)
plt.yticks(ticks = years)
plt.show()
```

### plt.scatter(x, y) - Scatter plot

```python
plt.scatter(years ,srk_revenue)
plt.xticks(ticks = years,rotation = 75)
plt.show()
```

### plt.pie(data) - Pie chart

```python
plt.figure(figsize=(12, 8))
plt.pie(srk_revenue, labels=srk_movie , autopct='%1.1f%%' )
plt.show()
```
