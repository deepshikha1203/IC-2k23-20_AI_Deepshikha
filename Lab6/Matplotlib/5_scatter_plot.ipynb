## Scatter Plot Properties
- ### `color` - to change color
- ### `size` - to change size
- ### `marker` - to change marker
- ### `transperancy` - to change opacity ( alpha )
- ### `colormap` - to give a gradient color

### `color` - to change color

```python
# For Single Color
colors = "red"

# For Multiple Colors , but has to pass no. of colors = no of elements
colors = ["red" , "pink" , "magenta" , "purple" ,"blue" ,"red" ,"pink" , "magenta" , "purple" ,"blue"]

# For Specific Function (List Comprehension)
colors = ["red" if x >= 130 else "green" for x in systolic_bp]

plt.scatter(age , systolic_bp , color = colors)
plt.xlabel("Age")
plt.xlabel("BP ( mmHg )")
plt.title("Age Vs BP")

plt.grid()

plt.show()
```

### `size` - to change size

```python
plt.scatter(age , systolic_bp , s = 10)
plt.xlabel("Age")
plt.xlabel("BP ( mmHg )")
plt.title("Age Vs BP")

plt.grid()

plt.show()
```

### `marker` - to change marker

```python
plt.scatter(age , systolic_bp , marker= "^")
plt.xlabel("Age")
plt.xlabel("BP ( mmHg )")
plt.title("Age Vs BP")

plt.grid()

plt.show()
```

### `transperancy` - to change opacity ( alpha )
- #### *Ranges Between 0 to 1* ,
- #### where 0 means invisible and 1 means visible and more inbetween

```python
plt.scatter(age , systolic_bp , alpha = 0.5)
plt.xlabel("Age")
plt.xlabel("BP ( mmHg )")
plt.title("Age Vs BP")

plt.grid()

plt.show()
```

### `colormap` - to give a gradient color
- #### **`cmap` to give colormap value**
- #### **`c` to give range**

```python
plt.scatter(age , systolic_bp , cmap = "cool" , c=systolic_bp)
plt.xlabel("Age")
plt.xlabel("BP ( mmHg )")
plt.title("Age Vs BP")

# used to show colorbar
plt.colorbar(label = "BP")

plt.grid()

plt.show()
```

## Annotate Points

```python
plt.scatter(age , systolic_bp , cmap = "cool" , c=systolic_bp)
plt.xlabel("Age")
plt.ylabel("BP ( mmHg )")
plt.title("Age Vs BP")

# used to show colorbar
plt.colorbar(label = "BP")

# To annotate data points
for i in range(len(names)):
    plt.annotate(text= names[i],
                 xy=(age[i] , systolic_bp[i]),
                 xytext=(age[i] + 1 , systolic_bp[i] - 0.6)
                )

plt.xlim(min(age) - 5 , max(age) + 10)
plt.ylim(min(systolic_bp) - 5 , max(systolic_bp) + 10)
plt.grid()

plt.show()
```
