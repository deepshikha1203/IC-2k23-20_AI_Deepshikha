# Getting Meaningfull Insights from Data

### Importing json Library to work with Json

```python
import json
```

### Function to Load Data from aa File

```python
def load_data(filename):
    with open(filename,'r') as f:
        data = json.load(f)
    return data
```

```python
data = load_data('new_data.json')
```

### Average of Ratings

```python
def avg_rating(data):
    totalRating = 0
    for user in data:
        totalRating += user['rating']

    return float(totalRating/len(data))
```

```python
avg_rating(data)
```

### No. of User Gave Rating Less Than 3

```python
def bad_rating_count(data):
    count = 0
    for user in data:
        if user['rating'] < 3:
            count += 1

    return count        
```

```python
bad_rating_count(data)
```
