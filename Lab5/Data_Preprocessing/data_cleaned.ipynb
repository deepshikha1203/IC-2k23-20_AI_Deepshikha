# Cleaning Data

### Importing json Library to work with Json

```python
import json
```

### Function to Load Data from aa File

```python
def load_data(filename):
    with open(filename,'r') as f:
        data = json.load(f)
    return data
```

```python
data = load_data('store_data.json')
```

### Implementing Data Cleaning Function

```python
def clean_data(data):
    rating_map = {'one':1,'two':2,'three':3,'four':4,'five':5}
    unique_data = set()
    cleaned_data = []
    
    for user in data:
        # change rating to numeric values [1-5]
        user_rating = str(user['rating']).strip().lower()
        if user_rating in rating_map:
            user_rating = rating_map[user_rating]

        user['rating'] = user_rating
        # print(user['rating'])


        # some data is missing - age
        if user.get('age') == None:
            user['age'] = None
        # print(user['age'])


        # removal of duplicate values
        current_user = user['name'].strip().lower()
        if current_user not in unique_data:
            unique_data.add(current_user)
            cleaned_data.append(user)
            
    return cleaned_data
```

```python
cleaned_data = clean_data(data)
```

### Saving Cleaned Data to a file

```python
with open('cleaned_data.json','w') as f:
    json.dump(cleaned_data,f)
```
