# Rule Based Recommending System

### Importing json Library to work with Json

```python
import json
```

### Function to Load Data from aa File

```python
def load_data(filename):
    with open(filename,'r') as f:
        data = json.load(f)
    return data
```

```python
data = load_data('new_data.json')
```

### Average of Ratings

```python
def recommendation(data):
    data_list = []

    for user in data :
        current = {}
        current['name'] = user['name']
        if user['rating'] < 4 :
            current['brand'] = "Samsung"
        else :
            current['brand'] = "Apple"

        data_list.append(current)  
    return data_list    
```

```python
recommendation(data)
```
