# Problem 3: Route Planning using Greedy Best-First Search

### Objective
Find a travel route from a start city to a destination city by picking the next city that seems closest according to a straight-line distance heuristic $h(n)$.

### How It Works
- Evaluation function: $f(n) = h(n)$
- A min-priority queue (`heapq`) extracts the node with the lowest heuristic cost at each step.

```python
import heapq

def greedy_route_planner(graph, start, goal, heuristics):
    # Priority queue stores tuples: (heuristic_val, current_city, path_taken)
    pq = [(heuristics[start], start, [start])]
    visited = set()
    
    while pq:
        h_val, current, path = heapq.heappop(pq)
        
        if current == goal:
            return path
            
        if current not in visited:
            visited.add(current)
            
            for neighbor in graph.get(current, []):
                if neighbor not in visited:
                    heapq.heappush(pq, (heuristics[neighbor], neighbor, path + [neighbor]))
                    
    return None

# Graph connections
city_network = {
    'City_A': ['City_B', 'City_C'],
    'City_B': ['City_D', 'City_E'],
    'City_C': ['City_F'],
    'City_D': ['Goal_City'],
    'City_E': ['Goal_City'],
    'City_F': ['Goal_City'],
    'Goal_City': []
}

# Straight-line distance estimates to Goal_City
h_values = {
    'City_A': 10,
    'City_B': 8,
    'City_C': 5,
    'City_D': 7,
    'City_E': 3,
    'City_F': 6,
    'Goal_City': 0
}

start = 'City_A'
destination = 'Goal_City'
route = greedy_route_planner(city_network, start, destination, h_values)

print("Optimal Planned Route:", " -> ".join(route))
```
